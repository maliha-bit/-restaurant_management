# Answer to the 1 no question

```python
number= list(map(int, input("enter list of numbers:").split()))
result=[]
for num in number:
  if num%4==0 and num%8!=0:
    result.append(num)

print(result)


```

**Answer to the 2 no question answer**

```python
def compute_ratio(x,y):
  try:
    return (x+y)/(x*y)
  except ZeroDivisionError:
    return "ERROR:Division not possible"

x,y=map(float, input("enter two numbers:").split())
result=compute_ratio(x,y)
print(result)
```

Answer to the 4 no question answer

```python
sentence=input("enter your sentence")
words=sentence.split()
result=[]
for vowel in words:
  if vowel[0].lower() in "aeiou":
    result.append(vowel)

print(result)
```
